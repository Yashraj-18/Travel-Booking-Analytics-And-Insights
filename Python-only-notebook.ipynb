# Travel Booking Analytics - Python Analysis

End-to-end exploratory data analysis and cancellation insights on hotel booking demand using Python (Pandas, NumPy).

### 1. Import Libraries

```python
import pandas as pd
import numpy as np
```

### 2. Load Dataset & Initial Exploration

```python
df = pd.read_csv('hotel_bookings.csv')
print(f'Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

### 3. Data Structure & Summary

```python
df.info()
```

### 4. Handling Missing Values
- Imputing missing numerical values for `children`, `agent`, `company` with 0
- Imputing missing `country` with 'Unknown'

```python
df['children'] = df['children'].fillna(0)
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
df['country'] = df['country'].fillna('Unknown')

# Verify no missing values remain in these columns
print('Missing values per column after imputation:')
print(df[['children', 'agent', 'company', 'country']].isnull().sum())
```

### 5. Overall & Hotel Type Cancellation Rates

```python
cancellation_rate = df['is_canceled'].mean() * 100
print(f'Overall Cancellation Rate: {cancellation_rate:.2f}%\n')

hotel_cancellation = df.groupby('hotel')['is_canceled'].mean() * 100
print('Cancellation Rate by Hotel Type:')
print(hotel_cancellation.round(2))
```

### 6. Lead Time Impact on Cancellations

```python
lead_time_summary = df.groupby('is_canceled')['lead_time'].mean().rename(index={0: 'Not Canceled', 1: 'Canceled'})
print('Average Lead Time (Days):')
print(lead_time_summary.round(2))
```

### 7. Customer Segmentation & Market Segment Analysis

```python
print('Cancellation Rate by Customer Type (%):')
print((df.groupby('customer_type')['is_canceled'].mean() * 100).sort_values(ascending=False).round(2))

print('\nCancellation Rate by Market Segment (%):')
print((df.groupby('market_segment')['is_canceled'].mean() * 100).sort_values(ascending=False).round(2))
```

### 8. Pricing (ADR) by Hotel Type

```python
print('Average Daily Rate (ADR) by Hotel (€):')
print(df.groupby('hotel')['adr'].mean().round(2))
```

### 9. Seasonality & Monthly Demand

```python
print('Total Bookings by Month of Arrival:')
print(df['arrival_date_month'].value_counts())
```

### 10. Room Type Pricing (ADR)

```python
print('Average Daily Rate (ADR) by Reserved Room Type (€):')
print(df.groupby('reserved_room_type')['adr'].mean().sort_values(ascending=False).round(2))
```
